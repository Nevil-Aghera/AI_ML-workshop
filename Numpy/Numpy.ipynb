#Broadcasting


```python
arr1 = np.array([[1], [2], [3]])  # Shape (3, 1)
arr2 = np.array([10, 20, 30])     # Shape (3,)

# Broadcasting over multiple axes
result = arr1 * arr2
print(result)
```
