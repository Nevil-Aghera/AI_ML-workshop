```python
tips_df =sns.load_dataset('tips')
tips_df
```

```python
sns.lineplot(x ='total_bill' , y = 'tip' , data=tips_df)
```

```python
sns.lineplot(
    x='total_bill',
    y='tip',
    data=tips_df,
    hue='day',
    style='day',
    palette='Set1',
    markers=True,
    dashes=False
)

plt.show()
```

```python
sns.lineplot(
    x='total_bill',
    y='tip',
    data=tips_df,
    hue='day',
    style='day',
    palette={
        'Thur': 'yellow',
        'Fri': 'green',
        'Sat': 'magenta',
        'Sun': 'gray'
    },
    markers=True,
    dashes=False
)

plt.show()
```
