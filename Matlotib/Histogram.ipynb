## Histogram

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
m1_students_age = np.random.randint(18,28,(100))
py_students_age = np.random.randint(14,28,(100))

```

```python
m1_students_age
```

```python
py_students_age
```

## ploting hist

```python
plt.hist(m1_students_age)
```

```python
plt.hist(m1_students_age)
plt.title('ml student age hist')
plt.xlabel('student age categary')
plt.ylabel('no . student age')

plt.show
```

```python
bins = [15,20,25,30,40,45]

plt.hist(m1_students_age , bins)
plt.title('ml student age hist')
plt.xlabel('student age categary')
plt.ylabel('no . student age')

plt.show
```

```python

plt.hist(m1_students_age ,rwidth=0.90, histtype = 'step' )
plt.title('ml student age hist')
plt.xlabel('student age categary')
plt.ylabel('no . student age')

plt.show
```

```python

plt.hist(m1_students_age ,rwidth=0.90, histtype = 'barstacked' )
plt.title('ml student age hist')
plt.xlabel('student age categary')
plt.ylabel('no . student age')

plt.show
```

```python

plt.hist(m1_students_age ,rwidth=0.90, histtype = 'barstacked', align='left' )
plt.title('ml student age hist')
plt.xlabel('student age categary')
plt.ylabel('no . student age')

plt.show
```

```python

plt.hist(m1_students_age ,rwidth=0.90, histtype = 'barstacked', align='left',orientation= 'horizontal' )
plt.title('ml student age hist')
plt.xlabel('student age categary')
plt.ylabel('no . student age')

plt.show
```

```python

plt.hist(
    m1_students_age,
    rwidth=0.90,
    histtype='barstacked',
    align='left',
    orientation='horizontal',
    color='purple',
    edgecolor='black'
)

plt.title('ml student age hist')    
plt.xlabel('student age categary')
plt.ylabel('no . student age')
plt.legend

plt.show
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(
    m1_students_age,
    rwidth=0.9,
    histtype='barstacked',
    align='left',
    orientation='horizontal',
    color='purple',
    edgecolor='black',
    label='ML Students'
)

plt.hist(
    py_students_age,
    rwidth=0.9,
    histtype='barstacked',
    align='left',
    orientation='horizontal',
    color='orange',
    edgecolor='black',
    label='Python Students'
)

plt.title('ML Student Age Histogram')
plt.xlabel('Number of Students')
plt.ylabel('Student Age Category')

# Show Legend
plt.legend(loc='upper right')

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(
    m1_students_age,
    rwidth=0.9,
    histtype='barstacked',
    align='left',
    orientation='horizontal',
    color='purple',
    edgecolor='black',
    label='ML Students'
)

plt.hist(
    py_students_age,
    rwidth=0.9,
    histtype='barstacked',
    align='left',
    orientation='vertical',
    color='orange',
    edgecolor='black',
    label='Python Students'
)

plt.title('ML Student Age Histogram')
plt.xlabel('Number of Students')
plt.ylabel('Student Age Category')

# Show Legend
plt.legend(loc='upper right')

plt.show()
```

```python
import matplotlib.pyplot as plt

bins = [15, 20, 25, 30, 35, 40, 45]

plt.hist(
    [m1_students_age, py_students_age],
    bins=bins,
    rwidth=0.9,
    color=['red', 'blue'],
    edgecolor='black',
    label=['ML Students', 'Py Students']
)

plt.title('ML & Python Student Age Histogram')
plt.xlabel('Student Age Category')
plt.ylabel('Number of Students')

plt.legend()

plt.show()
```
