## pie chart

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
classes = ['Python' , 'R' , 'Artificial Intelligence' , 'Machine Learning' , 'Data Science']
class1_students = [30, 10, 20, 25, 10]

plt.pie([1])
plt.show
```

```python
plt.pie(class1_students , labels=classes)
```

```python
explode = [0.1, 0, 0, 0, 0]

plt.pie(
    class1_students,
    labels=classes,
    explode= explode
)

plt.show()
```

```python
explode = (0.08, 0, 0, 0, 0)

colors = ['cyan', 'blue', 'red', 'gold', 'green']

textprops = {
    'fontsize': 10,
    'fontweight': 'bold',
    'color': 'black'
}

wedgeprops = {
    'linewidth': 2,
    'edgecolor': 'black'
}

plt.figure(figsize=(6, 4))  

plt.pie(
    class1_students,
    labels=classes,
    colors=colors,
    explode=explode,
    autopct='%1.1f%%',
    startangle=90,
    shadow=True,
    labeldistance=1.15,
    pctdistance=0.65,
    wedgeprops=wedgeprops,
    textprops=textprops
)

plt.title("Students Distribution", fontsize=16, fontweight='bold')

plt.legend(
    title="Courses",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)


plt.show()
```
