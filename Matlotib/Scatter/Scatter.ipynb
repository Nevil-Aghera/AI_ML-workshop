```python
df = pd.read_csv(r"F:\AI ml\Matlotib\Scatter\Data\googleplaystore_updated.csv" ,nrows=1000)
df
```

```python
x = df['Rating']
y = df['Reviews']

plt.title = ('Google Playstore')
plt.xlabel('Rating')
plt.ylabel('Reviews')
plt.scatter(x,y)

plt.show
```

```python
plt.scatter(
    x,
    y,
    c='brown',
    label='Google Play Store',
    marker='*',
    alpha=0.5,
    linewidths=5,
    edgecolors='brown'
)

plt.title=("Google Playstore")
plt.xlabel("Rating")
plt.ylabel("Reviews")

plt.legend()

plt.show()
```

```python
plt.scatter(
    x,
    y,
    c='brown',
    marker='*',
    label='Rating vs Reviews',
    alpha=0.5,
    linewidths=2,
    edgecolors='brown'
)

plt.scatter(
    x,
    df['Installs'],
    c='yellow',
    marker='o',
    label='Rating vs Installs',
    alpha=0.5,
    linewidths=2,
    edgecolors='black'
)

plt.title=('Google playstore ')
plt.xlabel('Rating')
plt.ylabel('Reviews')
plt.legend()
```
