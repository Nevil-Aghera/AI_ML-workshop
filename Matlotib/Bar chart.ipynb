```python
classes = ['Python', 'R', 'AI', 'ML', 'DS']

class1_students = [30, 10, 20, 25, 10]
class2_students = [40, 5, 20, 20, 10]
class3_students = [35, 5, 45, 8, 18]

plt.figure(figsize=(8, 8))

plt.bar(
    classes,
    class2_students,
    width=0.2,
    align='edge',
    color='c',
    edgecolor='b',
    linewidth=0.5,
    alpha=0.5,
    label = 'class2_students'
    
)

plt.bar(
    classes,
    class1_students,
    width=0.2,
    align='edge',
    color='gray',
    edgecolor='b',
    linewidth=0.5,
    alpha=0.5,
    label = 'class1_students'
    
)

plt.title("Students in Different Classes")
plt.xlabel("Courses")
plt.ylabel("Number of Students")

# Legend
plt.legend()


plt.show()
```

```python
classes_index = np.arange(len(classes))

width = 0.2

plt.figure(figsize=(8, 5))

# Class 1 Bar
plt.bar(
    classes_index,
    class1_students,
    width=width,
    color='skyblue',
    label='Class 1 Students'
)

# Class 2 Bar
plt.bar(
    classes_index + width,
    class2_students,
    width=width,
    color='gray',
    label='Class 2 Students'
)
plt.xticks(classes_index + width , classes , rotation = 20)

plt.title("Students in Different Classes")
plt.xlabel("Subjects")
plt.ylabel("Number of Students")


plt.legend()

plt.show()
```

```python
classes_index = np.arange(len(classes))

width = 0.2

plt.figure(figsize=(8, 5))

plt.bar(
    classes_index - width,
    class1_students,
    width=width,
    color='skyblue',
    label='Class 1 Students'
)

plt.bar(
    classes_index,
    class2_students,
    width=width,
    color='gray',
    label='Class 2 Students'
)

plt.bar(
    classes_index + width,
    class3_students,
    width=width,
    color='orange',
    label='Class 3 Students'
)

plt.xticks(classes_index, classes, rotation=20)

plt.title("Students in Different Classes")
plt.xlabel("Subjects")
plt.ylabel("Number of Students")

plt.legend()

plt.show()
```
