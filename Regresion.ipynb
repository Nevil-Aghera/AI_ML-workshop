```python
import pandas as pd
import numpy as np
import matplotlib as plt
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = {
    "Area": [1000, 1500, 2000, 2200, 2400, 2500],
    "Price": [30, 45, 60, 64, 66, 69]
}

df = pd.DataFrame(data)

print(df)

X = df[["Area"]]  # Independent Variable
```

```python
X = df[["Area"]]  # Independent Variable
y = df["Price"]   # Dependent Variable

# Step 4: Create Model
from sklearn.linear_model import LinearRegression

model = LinearRegression()

```

```python
# Step 5: Train Model
model.fit(X, y)
```

```python
# Step 6: Print Equation 
print("Slope:", model.coef_[0]) 
print("Intercept:", model.intercept_)
```

```python
# Step 7: Predict New Price
prediction = model.predict([[1800]])

print("Predicted Price:", prediction[0], "Lakhs")
```

```python
# Step 8: Predict Multiple Houses
new_area = [[1200], [1700], [2500]]

predictions = model.predict(new_area)

print(predictions)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(X, y, color="blue", label="Actual Data")

plt.plot(X, model.predict(X),
         color="red",
         linewidth=2,
         label="Regression Line")

plt.xlabel("Area (sq ft)")
plt.ylabel("Price (Lakhs)")
plt.title("Simple Linear Regression")
plt.legend()

plt.show()
```
